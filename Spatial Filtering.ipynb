***Averaging Grayscale Image using cv.filter2D***

```python
#Defining Averaging Kernel
avg_kernel=np.ones((3,3),np.float32)/9
im_avg_gray=cv.filter2D(im2,cv.CV_32F,avg_kernel)

#Plotting Images
fig,ax = plt.subplots(nrows=1, ncols=2,sharex='all',sharey='all', figsize=(10,8))
ax[0].imshow(X=im2,cmap='gray')
ax[0].set_title('Original Image')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(X=im_avg_gray,cmap='gray')
ax[1].set_title('After Average Filtering')
ax[1].set_xticks([])
ax[1].set_yticks([])
for a in ax:
    a.axis('off')
plt.show()
```

***Averaging Color Image using cv.filter2D***

```python
#Same Procedure can be done as cv.filter2D automatically filters all three channels independently
#Applying Average Kernel Defined in Previous Cell
im_avg_color=cv.filter2D(im1,cv.CV_32F,avg_kernel)
#Plotting Images
fig,ax = plt.subplots(nrows=1, ncols=2,sharex='all',sharey='all', figsize=(10,8))
ax[0].imshow(X=cv.cvtColor(im1,cv.COLOR_BGR2RGB))
ax[0].set_title('Original Image')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(X=cv.cvtColor(im_avg_color/255,cv.COLOR_BGR2RGB))
#using cv_CV32F generates floating point values in range 0-255
#However when imshow display fp32 it needs values in range of 0-1 Hence the extra'/255'
ax[1].set_title('After Average Filtering')
ax[1].set_xticks([])
ax[1].set_yticks([])
for a in ax:
    a.axis('off')
plt.show()
```

***Sobel Filtering Grayscale Image***

```python
#Defining Sobel Kernels
sobel_horizontal=np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
sobel_vertical=np.array([[-1,-2,-1],[0,0,0],[1,2,1]])

#Applying sobel kernels
im_sobel_horizontal_gray=cv.filter2D(im2,cv.CV_64F,sobel_horizontal)
im_sobel_vertical_gray=cv.filter2D(im2,cv.CV_64F,sobel_vertical)

#plotting images
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,16))
ax[0][0].imshow(X=im2,cmap='gray',vmin=0,vmax=255)
ax[0][0].set_title('Original Image')
ax[0][0].set_xticks([])
ax[0][0].set_yticks([])
ax[0][1].imshow(X=im_sobel_horizontal_gray,cmap='gray',vmin=0,vmax=255)
ax[0][1].set_title('Sobel Filtering')
ax[0][1].set_title('After Sobel.H Filtering')
ax[0][1].set_xticks([])
ax[0][1].set_yticks([])
ax[1][0].imshow(X=im2,cmap='gray',vmin=0,vmax=255)
ax[1][0].set_title('Original Image')
ax[1][0].set_xticks([])
ax[1][0].set_yticks([])
ax[1][1].imshow(X=im_sobel_vertical_gray,cmap='gray',vmin=0,vmax=255)
ax[1][1].set_title('After Sobel.V Filtering')
ax[1][1].set_xticks([])
ax[1][1].set_yticks([])
for a in ax:
    for b in a:
        b.axis('off')
plt.show()
```

***Sobel Filtering Color Image***

```python
#Applying sobel kernels
im_sobel_horizontal_color=cv.filter2D(im1,cv.CV_64F,sobel_horizontal)
im_sobel_vertical_color=cv.filter2D(im1,cv.CV_64F,sobel_vertical)

im_sobel_horizontal_display = cv.convertScaleAbs(im_sobel_horizontal_color)
im_sobel_vertical_display = cv.convertScaleAbs(im_sobel_vertical_color)

#plotting images
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,16))

ax[0][0].imshow(cv.cvtColor(im1, cv.COLOR_BGR2RGB))
ax[0][0].set_title('Original Image')

ax[0][1].imshow(cv.cvtColor(im_sobel_horizontal_display, cv.COLOR_BGR2RGB))
ax[0][1].set_title('After Sobel H Filtering')

ax[1][0].imshow(cv.cvtColor(im1, cv.COLOR_BGR2RGB))
ax[1][0].set_title('Original Image')

ax[1][1].imshow(cv.cvtColor(im_sobel_vertical_display, cv.COLOR_BGR2RGB))
ax[1][1].set_title('After Sobel V Filtering')

for row in ax:
    for a in row:
        a.axis('off')

plt.show()
```
