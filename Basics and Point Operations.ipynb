**Image Creation**

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

```python
#initializing a 6x8 numpy array and setting one pixel as 255
im = np.zeros((6,8),dtype=np.uint8) #unsigned integers of 8 bits
im[2,3]=255
```

```python
#displaying image
fig,ax=plt.subplots(nrows=1,ncols=1, figsize=(8,6))
ax.imshow(X=im, cmap='gray', vmin=0, vmax=255)
ax.xaxis.set_ticks_position(position='top')
plt.show()
```

```python
#A three dimentional numpy array used to depict RGB color space
im_ = np.zeros((6,8,3),dtype=np.uint8)
im_[2,3]=[255,0,0] #red
im_[2,4]=[0,255,0] #green
im_[2,5]=[0,0,255] #blue
```

```python
#plotting 3-d image
fig, ax = plt.subplots(nrows=1,ncols=1, figsize=(8,6))
ax.imshow(X=im_)
ax.xaxis.set_ticks_position(position='top')
plt.show()
```

**Increasing Brightness**

```python
#Taking a grayscale image and adding a value of 100 to each pixel
im2=cv.imread('D:\EN3160 - Image Procesing and Machine Vision Codes\images\emma_gray.jpg', cv.IMREAD_GRAYSCALE)
assert im2 is not None
im3=cv.add(im2, 100)

#Plotting the two images
fig,ax=plt.subplots(nrows=1,ncols=2, figsize=(10,6))
ax[0].imshow(X=im2,cmap='gray',vmin=0,vmax=255)
ax[0].set_title('Original_Image')
ax[1].imshow(X=im3,cmap='gray',vmin=0,vmax=255)
ax[1].set_title('Brightness_Increased_Image')
for a in ax:
    a.axis('off')
plt.show()
```

**Extracting Color Channels From Images**

```python
im5=cv.imread('D:\EN3160 - Image Procesing and Machine Vision Codes/images/flowers.jpg')
assert im5 is not None
im5_blue = im5.copy()
im5_blue[:,:,1],im5_blue[:,:,2]=0,0; #G,R channels turned 0
fig,ax=plt.subplots(nrows=1,ncols=2, figsize=(10,6))
ax[0].imshow(cv.cvtColor(im5, cv.COLOR_BGR2RGB))
ax[0].set_title('Original_Image')
ax[1].imshow(cv.cvtColor(im5_blue,cv.COLOR_BGR2RGB))
ax[1].set_title('Blue Channel')
for a in ax:
    a.axis('off')
plt.show()
```

**Intensity Transformations**

```python
#Consider doing an identity intensity transformation on im2
t=np.array(object=[i for i in range(256)], dtype=np.uint8)
#applying transformation
im6=t[im2]
#plotting transformation curve, original image and transformed image
fig,ax=plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(im2,cmap='gray',vmin=0,vmax=255)
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].imshow(im2,cmap='gray',vmin=0,vmax=255)
ax[1].set_title('Transformed_Image')
ax[1].axis('off')
ax[2].plot(t,color='black')
ax[2].set_title('Identity_Transformation')
ax[2].axis('equal')
plt.show()
```

**Negative Transformation**

```python
im7=cv.imread('D:\EN3160 - Image Procesing and Machine Vision Codes/images/will_smith.jpg',cv.IMREAD_GRAYSCALE)
assert im7 is not None

#Defining Transformation
t1=np.array(object=[255-i for i in range(256)], dtype=np.uint8)
#applying transformation
im7_=t1[im7]
#plotting original image, tranformation, transformed image
fig,ax=plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(X=im7,cmap='gray',vmin=0,vmax=255)
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].plot(t1,color='black')
ax[1].set_title('Negative_Transformation')
ax[2].imshow(im7_,cmap='gray',vmin=0,vmax=255)
ax[2].set_title('Transformed_Image')
ax[2].axis('off')
plt.show()
```

**Intensity Windowing**

```python
c=np.array([(100,50), (150,200)])

t1_ = np.linspace(0,c[0,1],c[0,0]+1-0).astype('uint8')
t2_ = np.linspace(c[0,1]+1,c[1,1],c[1,0]-c[0,0]).astype('uint8')
t3_ = np.linspace(c[1,1]+1,255,255-c[1,0]).astype('uint8')

transform = np.concatenate((t1_,t2_),axis=0).astype('uint8')
transform = np.concatenate((transform,t3_),axis=0).astype('uint8')

im8=transform[im5]

fig,ax = plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(cv.cvtColor(im5,cv.COLOR_BGR2RGB))
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].plot(transform,color='black')
ax[1].set_title('Transformation')
ax[2].imshow(cv.cvtColor(im8,cv.COLOR_BGR2RGB))
ax[2].set_title('Transformed_Image')
ax[2].axis('off')
plt.show()
```

**Gamma Correction**

```python
# For 0<gamma<1 maps narrow range of dark pixels to a wider range
# For gamma=1 it is a Identity Transformation
# For 1<gamma maps narrow range of bright pixels to wider range of bright pixels

#reading image
im10=cv.imread('D:\EN3160 - Image Procesing and Machine Vision Codes/images/overexposed_city.jpg')
assert im10 is not None
#defining gamma transformation for gamma>1
t4=np.array([(i/255.0)**(3)*255 for i in np.arange(0, 256)]).astype(np.uint8)
#applying transformation
im10_=cv.LUT(im10,t4)
#plotting
fig,ax = plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(cv.cvtColor(im10,cv.COLOR_BGR2RGB))
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].plot(t4,color='black')
ax[1].set_title('Transformation')
ax[2].imshow(cv.cvtColor(im10_,cv.COLOR_BGR2RGB))
ax[2].set_title('Transformed_Image')
ax[2].axis('off')
plt.show()
```

```python
#defining gamma transformation for gamma = 1
t5=np.array([(i/255.0)**(1)*255 for i in np.arange(0, 256)]).astype(np.uint8)
#applying transformation
im11=cv.LUT(im10,t5)
#plotting
fig,ax = plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(cv.cvtColor(im10,cv.COLOR_BGR2RGB))
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].plot(t5,color='black')
ax[1].set_title('Transformation')
ax[2].imshow(cv.cvtColor(im11,cv.COLOR_BGR2RGB))
ax[2].set_title('Transformed_Image')
ax[2].axis('off')
plt.show()
```

```python
#defining gamma transformation for gamma < 1
t6=np.array([(i/255.0)**(0.3)*255 for i in np.arange(0, 256)]).astype(np.uint8)
#applying transformation
im12=cv.LUT(im10,t5)
#plotting
fig,ax = plt.subplots(nrows=1,ncols=3, figsize=(12,4))
ax[0].imshow(cv.cvtColor(im10,cv.COLOR_BGR2RGB))
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].plot(t6,color='black')
ax[1].set_title('Transformation')
ax[2].imshow(cv.cvtColor(im12,cv.COLOR_BGR2RGB))
ax[2].set_title('Transformed_Image')
ax[2].axis('off')
plt.show()
```

**Histogram Equalization**

```python
#reading image
im13=cv.imread('D:\EN3160 - Image Procesing and Machine Vision Codes/images/overexposed_city.jpg',cv.IMREAD_GRAYSCALE)
assert im13 is not None
#histogram equalization
im14 = cv.equalizeHist(src=im13)
#plotting
fig,ax = plt.subplots(nrows=1,ncols=2, figsize=(8,4))
ax[0].imshow(im13,cmap='gray',vmin=0,vmax=255)
ax[0].set_title('Original_Image')
ax[0].axis('off')
ax[1].imshow(im14,cmap='gray',vmin=0,vmax=255)
ax[1].set_title('Histogram Equalized Image')
ax[1].axis('off')
plt.show()
```
